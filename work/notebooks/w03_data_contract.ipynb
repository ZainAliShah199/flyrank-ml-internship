I will use a mid-panel month (2026-03) for analysis because it represents historical data and avoids using the final month as a test period.

## Fields

### Features (used for ML signals)
- impressions_90d: Shows page visibility and search exposure.
- ctr: Shows how effectively impressions turn into clicks.
- avg_position: Shows search ranking position.
- content_age_days: Shows how old the content is.
- days_since_last_update: Shows freshness of the page.

### Label / Proxy
- trend_direction: Used as a proxy label for whether a page is declining or improving.

### Context fields
- month: Defines the reporting period.
- content_type: Helps compare different content groups.
- client_id: Used only for validation grouping if required.

### Excluded fields
- trend_pct: Excluded because it is directly related to the outcome and can create data leakage.
- Any future outcome fields: excluded because they would not be available at the decision moment.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Dataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())


# Row count
print("\nRows in dataset:")
print(len(df))


# Time window check
if "month" in df.columns:
    print("\nAvailable months:")
    print(df["month"].unique())


# Missing values
print("\nMissing values:")
print(df.isna().sum().sort_values(ascending=False).head(10))


# Check duplicate page-month records
possible_keys = [c for c in ["url", "page", "page_id", "month"] if c in df.columns]

print("\nPossible key columns:", possible_keys)
```

## Data limitations

This dataset can support decision-making about which pages may deserve review, but it cannot explain every reason behind search performance changes.

The data cannot prove causation or explain Google's ranking algorithm.

Limitations include:

- Historical data may not represent all future search behavior.
- Some early rows may contain only Search Console information and may lack complete context.
- Time windows may overlap, which can affect comparisons between periods.
- Observations show patterns and relationships, not guaranteed outcomes.

The results should be treated as measured, directional, and decision-support only.
