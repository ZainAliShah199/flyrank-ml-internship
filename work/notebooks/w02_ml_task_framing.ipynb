## 2. Target or proxy

The target is whether a content page is declining in search performance. In the starter dataset, I will use **trend_direction = "down"** as a proxy label for declining pages. This is an observed outcome based on historical search performance rather than a manually assigned opinion. The model will learn patterns from observable features such as impressions, click-through rate, content age, and average position to estimate which pages should be prioritized for review.

## 3. Success metric

The primary success metric is **Precision@50**. This measures how many of the top 50 pages recommended by the model are actually declining. A higher Precision@50 means the ranking is more useful because it helps content teams spend their limited time reviewing pages that are more likely to need attention. This metric aligns well with the practical goal of prioritizing work rather than evaluating every page equally.

## 5. Why ML beats a fixed rule here

A fixed rule usually depends on one or two manually selected conditions, such as page age or impressions. However, page performance is influenced by many factors working together, including impressions, click-through rate, average position, content age, word count, and engagement metrics. Machine learning can combine these signals to identify patterns that are difficult to capture with a simple if/else rule. The model provides decision support by producing a ranked list of pages, while the final decision remains with the content team.
